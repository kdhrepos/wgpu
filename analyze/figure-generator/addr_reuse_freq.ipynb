```python
import pandas as pd
import numpy as np

# 1. 시간 정규화 및 interval 계산 (기존 로직 유지)
# ns = 1e-9, time_interval = 10000
t_min = trace_df['time'].min()
time_vals = trace_df['time'].to_numpy()
trace_df['interval'] = ((time_vals - t_min) / ns // time_interval) * time_interval

# =========================================================================
# 지표 1: 특정 interval 내에서 [PC가 특정 addr에] 몇 번 접근했는지
# =========================================================================
interval_pc_addr_stats = trace_df.groupby(['interval', 'pc', 'addr'], as_index=False).agg(
    access_count_in_interval=('access_count', 'sum')
)

# =========================================================================
# 지표 2: 해당 [PC-addr 쌍]이 전체 시간 동안 총 몇 개의 interval에서 등장했는지 (수명)
# =========================================================================
pc_addr_lifespan = trace_df.groupby(['pc', 'addr'], as_index=False).agg(
    active_intervals_count=('interval', 'nunique')
)

# =========================================================================
# 3. 두 지표 병합
# =========================================================================
# interval 통계에 lifespan(수명) 정보를 병합합니다.
result_df = interval_pc_addr_stats.merge(pc_addr_lifespan, on=['pc', 'addr'], how='left')

# 정렬: 여러 interval에 등장한 수명이 긴 쌍부터 -> PC별로 묶어서 -> 시간 순으로
result_df = result_df.sort_values(by=['interval'])
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.ticker as ticker
import os

def plot_addr_mean_access_by_pc(df, result_dir=".", benchmark="benchmark", pc_color_map=None, pc_marker_map=None):
    """
    x축: addr (메모리 주소)
    y축: mean_access_per_interval (해당 주소의 등장 구간 내 평균 접근 횟수)
    범례(색상/마커): PC
    
    * 주의: 입력되는 df는 사전에 pc, addr별로 mean_access_per_interval이 
            계산되어 있어야 합니다.
    """
    
    # ---------------------------------------------------------
    # 1. 시각화용 데이터 준비
    # ---------------------------------------------------------
    # 외부에서 계산된 데이터를 복사하여 사용
    plot_df = df[['pc', 'addr', 'mean_access_per_interval']].copy()
    
    # x축 시각화를 위해 16진수 문자열을 정수로 고속 변환
    if plot_df['addr'].dtype == 'O':
        plot_df['addr_int'] = plot_df['addr'].apply(int, base=16)
    else:
        plot_df['addr_int'] = plot_df['addr']

    # ---------------------------------------------------------
    # 2. 글로벌 스타일 설정 (논문용)
    # ---------------------------------------------------------
    plt.rcParams['figure.dpi'] = 300
    plt.rcParams['savefig.dpi'] = 300
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['axes.axisbelow'] = True
    
    fig, ax = plt.subplots(figsize=(10, 5))
    
    ax.set_facecolor('white')
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    # 접근 횟수는 편차가 클 수 있으므로 그리드를 좀 더 촘촘히 그립니다.
    ax.grid(True, axis='y', linestyle='--', alpha=0.6, color='lightgray')

    # ---------------------------------------------------------
    # 3. 색상 및 마커 (Marker) 맵핑 일관성 보장
    # ---------------------------------------------------------
    unique_pcs = sorted(plot_df['pc'].unique().tolist())
    
    # 외부에서 맵핑 정보가 안 들어오면 자체적으로 오름차순 고정 생성 (단일 벤치마크 일관성 보장)
    if pc_color_map is None:
        palette = sns.color_palette("husl", len(unique_pcs))
        pc_color_map = dict(zip(unique_pcs, palette))
        
    if pc_marker_map is None:
        markers = ['o', 's', '^', 'D', 'v', 'p', '*', 'h', '<', '>'] 
        pc_marker_map = {pc: markers[i % len(markers)] for i, pc in enumerate(unique_pcs)}
    
    # ---------------------------------------------------------
    # 4. PC별로 반복하며 Scatter Plot 그리기 (Groupby 최적화)
    # ---------------------------------------------------------
    grouped_df = plot_df.groupby('pc')
    
    for pc in unique_pcs:
        # 그룹에 데이터가 없는 경우를 대비
        if pc not in grouped_df.groups:
            continue
            
        pc_data = grouped_df.get_group(pc)
        
        color = pc_color_map.get(pc, '#808080')
        marker = pc_marker_map.get(pc, 'o')
        
        label_str = f"{pc}" if isinstance(pc, (int, np.integer)) else f"{pc}"
        
        ax.scatter(
            pc_data['addr_int'], 
            pc_data['mean_access_per_interval'], 
            color=color, 
            marker=marker, 
            label=label_str,
            alpha=0.7,      
            edgecolors='white', 
            linewidth=0.5,
            s=25  # 마커 크기
        )

    # ---------------------------------------------------------
    # 5. 축(Axis) 및 라벨 디자인
    # ---------------------------------------------------------
    ax.set_xlabel('Memory Address (Hex)', fontsize=11, fontweight='bold', labelpad=10)
    ax.set_ylabel('Mean Access Count per Interval', fontsize=11, fontweight='bold', labelpad=10)
    ax.set_title(f'[{benchmark}] Mean Access Count by Generating PC', fontsize=13, fontweight='bold', pad=15)
    
    # 💡 데이터 편차가 극단적으로 크면 Y축을 로그 스케일로 보는 것이 좋을 수 있습니다.
    # 만약 편차가 너무 커서 바닥에 다 붙어 나온다면 주석을 해제하세요.
    # ax.set_yscale('log')

    # X축 16진수 포맷팅 및 개수 제한
    ax.xaxis.set_major_locator(ticker.MaxNLocator(nbins=15))
    def hex_formatter(x, pos):
        return f"0x{int(x):X}"
    ax.xaxis.set_major_formatter(ticker.FuncFormatter(hex_formatter))

    plt.setp(ax.get_xticklabels(), rotation=45, ha='right', fontsize=9)
    plt.setp(ax.get_yticklabels(), fontsize=9)

    # ---------------------------------------------------------
    # 6. 범례(Legend) 설정 (우측 바깥)
    # ---------------------------------------------------------
    ax.legend(
        loc='upper left', 
        bbox_to_anchor=(1.02, 1), 
        title='Program Counter (PC)', 
        title_fontsize=10,
        fontsize=9,
        frameon=True,
        edgecolor='lightgray',
        markerscale=1.5
    )

    fig.tight_layout()

    # ---------------------------------------------------------
    # 7. 저장
    # ---------------------------------------------------------
    os.makedirs(result_dir, exist_ok=True)
    out_file = f"{result_dir}/{benchmark}_addr_mean_access_scatter.png"
    
    fig.savefig(out_file, dpi=300, bbox_inches='tight', transparent=False)
    plt.close(fig)

    print(f"✅ Saved figure: {out_file}")
```

```python
def plot_addr_lifespan_by_pc(df, result_dir=".", benchmark="benchmark"):
    """
    x축: addr (메모리 주소)
    y축: active_intervals_count (주소가 살아남은 interval 개수)
    범례(색상/마커): PC
    """
    
    # ---------------------------------------------------------
    # 1. 시각화용 데이터 준비
    # ---------------------------------------------------------
    # 원본 데이터에 NaN이 있을 경우를 대비하여 복사본을 안전하게 만듭니다.
    plot_df = df[['interval', 'pc', 'addr', 'active_intervals_count']].drop_duplicates().copy()
    
    # 💡 최적화: apply(lambda) 대신 내장 int 함수의 base=16 활용 (속도 극대화)
    if plot_df['addr'].dtype == 'O':
        plot_df['addr_int'] = plot_df['addr'].apply(int, base=16)
    else:
        plot_df['addr_int'] = plot_df['addr']

    # ---------------------------------------------------------
    # 2. 논문급 고품질 글로벌 스타일 설정
    # ---------------------------------------------------------
    plt.rcParams['figure.dpi'] = 300
    plt.rcParams['savefig.dpi'] = 300
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['axes.axisbelow'] = True
    
    fig, ax = plt.subplots(figsize=(10, 5))
    
    ax.set_facecolor('white')
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.grid(True, axis='y', linestyle='--', alpha=0.5, color='gray')

    # ---------------------------------------------------------
    # 3. 색상 및 마커 (Marker) 맵핑
    # ---------------------------------------------------------
    unique_pcs = plot_df['pc'].unique()
    palette = sns.color_palette("tab20", n_colors=len(unique_pcs))
    
    # 논문 인쇄용 대비가 좋은 마커 리스트
    markers = ['o', 's', '^', 'D', 'v', 'p', '*', 'h', '<', '>'] 
    
    # ---------------------------------------------------------
    # 4. PC별로 반복하며 Scatter Plot 그리기 (Groupby 활용 최적화)
    # ---------------------------------------------------------
    # 💡 최적화: 매번 필터링하지 않고 groupby를 통해 순회
    grouped_df = plot_df.groupby('pc')
    
    for i, pc in enumerate(unique_pcs):
        pc_data = grouped_df.get_group(pc)
        
        color = palette[i]
        marker = markers[i % len(markers)]
        
        label_str = f"{pc}" if isinstance(pc, (int, np.integer)) else f"{pc}"
        
        ax.scatter(
            pc_data['addr_int'], 
            pc_data['active_intervals_count'], 
            color=color, 
            marker=marker, 
            label=label_str,
            alpha=0.7,      
            edgecolors='white', 
            linewidth=0.5,
            s=20 # 💡 마커 크기를 10에서 20으로 소폭 상향 (가시성 확보)
        )

    # ---------------------------------------------------------
    # 5. 축(Axis) 및 라벨 디자인
    # ---------------------------------------------------------
    ax.set_xlabel('Memory Address (Hex)', fontsize=11, fontweight='bold', labelpad=10)
    ax.set_ylabel('Active Intervals Count (Lifespan)', fontsize=11, fontweight='bold', labelpad=10)
    ax.set_title(f'[{benchmark}] Address Lifespan by Generating PC', fontsize=13, fontweight='bold', pad=15)

    max_interval_count = plot_df['interval'].nunique()
    ax.set_ylim(0, max_interval_count * 1.05) 
    
    # 💡 가독성 개선: x축에 주소값(눈금)이 너무 많이 찍히는 것을 방지
    ax.xaxis.set_major_locator(ticker.MaxNLocator(nbins=15))
    
    def hex_formatter(x, pos):
        return f"0x{int(x):X}"
    ax.xaxis.set_major_formatter(ticker.FuncFormatter(hex_formatter))

    plt.setp(ax.get_xticklabels(), rotation=45, ha='right', fontsize=9)
    plt.setp(ax.get_yticklabels(), fontsize=9)

    # ---------------------------------------------------------
    # 6. 범례(Legend) 설정
    # ---------------------------------------------------------
    # 범례의 점이 너무 작아 안 보일 수 있으므로 markerscale 옵션 추가
    ax.legend(
        loc='upper left', 
        bbox_to_anchor=(1.02, 1), 
        title='Program Counter (PC)', 
        title_fontsize=10,
        fontsize=9,
        frameon=True,
        edgecolor='lightgray',
        markerscale=1.5 # 💡 범례 안의 마커 크기 확대
    )

    fig.tight_layout()

    # ---------------------------------------------------------
    # 7. 저장
    # ---------------------------------------------------------
    os.makedirs(result_dir, exist_ok=True)
    out_file = f"{result_dir}/{benchmark}_addr_lifespan_scatter.png"
    
    fig.savefig(out_file, dpi=300, bbox_inches='tight', transparent=False)
    plt.close(fig)

    print(f"✅ Saved figure: {out_file}")
```
