```python
base_dir=f'{username}/mgvm/results/reports'
sub_dir = "20260513-124102-direct-6x6"
platform = "direct"
report_dir = f'{base_dir}/{sub_dir}'

xDim=6
yDim=6

"""
kernel time, simulation time
"""
completed = 0

print(f"Benchmark\tTotal Kernel Time\tSimulation Hours")

for benchmark in benchmarks:
    report_file = f"{report_dir}/{benchmark}_{platform}_{xDim}x{yDim}.csv"
    
    if not os.path.exists(report_file):
        # print(f"{benchmark}")
        # print(f"  Skip: No file {report_file}")
        continue

    df = pd.read_csv(
        report_file,
        skipinitialspace=True,
        on_bad_lines='skip' # 오류 라인(열 개수가 안 맞는 줄)을 무시합니다.
    )

    try:
        # .item() 대신 .iloc[0]을 쓰면 중복 값이 있을 때 발생할 수 있는 에러도 피할 수 있습니다.
        total_kernel_time = df[df['what']=='total-kernel-time']['value'].iloc[0]
        total_kernel_time = float(total_kernel_time)

        simulation_time = df[df['what']=='simulation_time']['value'].iloc[0]
        simulation_time = float(simulation_time)
        simulation_hours = simulation_time / 3600
    except IndexError:
        # 지표가 비어있는 경우를 안전하게 처리합니다.
        total_kernel_time = 0.0
        simulation_hours = 0.0

    if total_kernel_time > 0:
        completed+=1
        
    print(f"{benchmark}\t{total_kernel_time}\t{simulation_hours}")

print(f"Completed: {completed}/{len(benchmarks)}")
```

```python
"""
L1 Hit Rate
"""
report_dir = f'{base_dir}/{platform}'

print("benchmark\tL1 Hit Rate")
for benchmark in benchmarks:
    report_file = f"{report_dir}/{benchmark}_{platform}_6x6.csv"
    
    if not os.path.exists(report_file):
        print(f"  Skip: No file {report_file}")
        continue

    df = pd.read_csv(
        report_file,
        skipinitialspace=True,
    )

    total_l1_read_hit = df[df['what']=='total-l1-read-hit']['value'].item()
    total_l1_read_miss = df[df['what']=='total-l1-read-miss']['value'].item()
    total_l1_read_mshr_hit = df[df['what']=='total-l1-read-mshr-hit']['value'].item()
    total_l1_write_hit = df[df['what']=='total-l1-write-hit']['value'].item()
    total_l1_write_miss = df[df['what']=='total-l1-write-miss']['value'].item()
    total_l1_write_mshr_hit = df[df['what']=='total-l1-write-mshr-hit']['value'].item()

    total_l1_hit = total_l1_read_hit + total_l1_read_mshr_hit + total_l1_write_hit + total_l1_write_mshr_hit
    total_transactions = total_l1_read_hit + total_l1_read_miss + total_l1_read_mshr_hit + \
                        total_l1_write_hit + total_l1_write_miss + total_l1_write_mshr_hit
    l1_hit_rate = (total_l1_hit / total_transactions)
    print(f"{benchmark}\t{l1_hit_rate}")
```
