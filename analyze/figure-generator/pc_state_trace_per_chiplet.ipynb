```python
benchmarks = [
    # Low-sharing Apps
    "fdtd2d", "jacobi1d", "fastwalshtransform", "relu", "fft", "fir", 
    "sssp", "convolution2d", "color", "bitonicsort", "pagerank", "simpleconvolution",
    # High-sharing Apps
    'gups', 'syr2k', 'spmv', 'mis', 'stencil2d', 'gemm', 'lu', 
    'matrixmultiplication', 'mm2', 'mm3', 'gaussian', 'matrixtranspose', 
    'im2col', 'vgg16', 'bfs',
]
```

```python
base_dir=f'{username}/mgvm/results/pc-state-traces'
date = "20260317-003053-staticdist-9"
trace_dir = f"{base_dir}/{date}"
print(f"{base_dir}/{date}")
```

```python
cluster_size = 9
dist_type = 'staticdist'
```

```python
for benchmark in benchmarks:
    trace_file = f"{trace_dir}/{benchmark}_{dist_type}_{cluster_size}_6x6_pc_state_trace.csv"
    print(trace_file)
    
    if not os.path.exists(trace_file):
        print(f"{benchmark} not exists") 
        continue

    df = pd.read_csv(trace_file)
    df = df[df['pc'] != 0]
    df_sorted = df.sort_values(by='time', ascending=False)
    df_latest = df_sorted.drop_duplicates(subset=['chiplet', 'pc'], keep='first')
    df_latest = df_latest.sort_values(by=['chiplet', 'time']).reset_index(drop=True)
    
    result_dir = f"./latest-only"
    os.makedirs(f"{result_dir}", exist_ok=True)

    df_latest.to_csv(f"{result_dir}/{benchmark}_latest_pc_only.csv", index=False)
```
