```python
base_dir=f'{username}/mgvm/results/access-counts'
result_dir=f'{username}/mgvm/results/figs/access-counts'
print(base_dir)
print(result_dir)
```

```python
for benchmark in benchmarks:
    input = f'{base_dir}/20260130-040704/{benchmark}_mesh_6x6_access_count.csv'

    df = pd.read_csv(input)

    plt.figure(figsize=(10, 6)) 
    plt.plot(df[' address'], df[' count'], 'o', label='Count per Address', markersize=4)

    plt.title('Access count per address', fontsize=14)
    plt.xlabel('Address', fontsize=12)
    plt.ylabel('Count', fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.6)

    plt.ticklabel_format(style='plain', axis='x')
    plt.xticks(rotation=45)

    plt.legend()
    plt.tight_layout()

    plt.savefig(f'{result_dir}/{benchmark}.png', dpi=300)
```
